# 🏠 House Price Prediction

## Week 9 - Introduction to Machine Learning

This project develops a machine learning model to predict house prices
using property-related features such as area, bedrooms, and location.

### Objectives

- Understand basic machine learning concepts
- Explore and clean the house price dataset
- Prepare numerical and categorical features
- Implement Linear Regression from scratch
- Build regression models using scikit-learn
- Evaluate models using MAE, MSE, RMSE, and R²
- Compare multiple regression algorithms
- Visualize actual versus predicted prices
- Identify important features affecting house prices
